```python
# Fill missing numerical values with median
for col in df.select_dtypes(include=np.number).columns:
    df[col] = df[col].fillna(df[col].median())

print("Missing values after preprocessing:")
print(df.isnull().sum())
```

```python
median_production = df["Energy_Production_MWh"].median()

print("Median Energy Production:", median_production)
```

```python
df["Energy_Availability"] = np.where(
    df["Energy_Production_MWh"] >= median_production,
    "High",
    "Low"
)

print(df["Energy_Availability"].value_counts())
```

```python
plt.figure(figsize=(7, 5))

sns.countplot(
    x="Energy_Availability",
    data=df
)

plt.title("High vs Low Renewable Energy Availability")
plt.xlabel("Energy Availability")
plt.ylabel("Number of Records")
plt.show()
```

```python
features = [
    "Type_of_Renewable_Energy",
    "Installed_Capacity_MW",
    "Energy_Consumption_MWh",
    "Energy_Storage_Capacity_MWh",
    "Storage_Efficiency_Percentage",
    "Grid_Integration_Level",
    "Initial_Investment_USD",
    "Funding_Sources",
    "Financial_Incentives_USD",
    "GHG_Emission_Reduction_tCO2e",
    "Air_Pollution_Reduction_Index",
    "Jobs_Created"
]

X = df[features]

y = df["Energy_Availability"]

print("Feature shape:", X.shape)
print("Target shape:", y.shape)
```

```python
X_train, X_test, y_train, y_test = train_test_split(
    X,
    y,
    test_size=0.20,
    random_state=42,
    stratify=y
)

print("Training data:", X_train.shape)
print("Testing data:", X_test.shape)
```

```python
model = DecisionTreeClassifier(
    criterion="entropy",
    max_depth=5,
    random_state=42
)

model.fit(X_train, y_train)

print("Decision Tree model trained successfully")
```

```python
y_pred = model.predict(X_test)

print("Predictions:")
print(y_pred[:20])
```

```python
accuracy = accuracy_score(y_test, y_pred)

print("Accuracy:", accuracy)
print("Accuracy (%):", accuracy * 100)
```

```python
precision = precision_score(
    y_test,
    y_pred,
    pos_label="High"
)

recall = recall_score(
    y_test,
    y_pred,
    pos_label="High"
)

f1 = f1_score(
    y_test,
    y_pred,
    pos_label="High"
)

print("Precision:", precision)
print("Recall:", recall)
print("F1 Score:", f1)
```

```python
print("Classification Report:")
print(classification_report(y_test, y_pred))
```

```python
cm = confusion_matrix(
    y_test,
    y_pred,
    labels=["Low", "High"]
)

plt.figure(figsize=(7, 5))

sns.heatmap(
    cm,
    annot=True,
    fmt="d",
    cmap="Blues",
    xticklabels=["Low", "High"],
    yticklabels=["Low", "High"]
)

plt.title("Confusion Matrix - Decision Tree")
plt.xlabel("Predicted")
plt.ylabel("Actual")
plt.show()
```

```python
importance = pd.DataFrame({
    "Feature": X.columns,
    "Importance": model.feature_importances_
})

importance = importance.sort_values(
    by="Importance",
    ascending=False
)

display(importance)
```
